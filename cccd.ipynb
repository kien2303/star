```python
import numpy as np
import matplotlib.pyplot as plt
import pytesseract
from PIL import Image
filename='/content/z6565578644575_e2368584ef6d14fcf9020d037c0e3147.jpg'
img=plt.imread(filename)
plt.imshow(img)
plt.show()
img1=np.array(Image.open(filename))
text=pytesseract.image_to_string(img1,lang='vie')
print(text)
split_text=text.split('\n')
print(split_text)
cleaned_text = [item.strip() for item in split_text if item.strip()]
print(cleaned_text)
```

```python
import openpyxl

wb = openpyxl.Workbook()
sht = wb.active

# Iterate through the split_text list and write each element to a separate row in column A
for i, line in enumerate(cleaned_text):
    # Replace illegal characters with an empty string
    line = line.replace('\x0c', '')
    sht.cell(row=i + 1, column=1).value = line  # i + 1 to start from row 1

wb.save('my_workbook.xlsx')
```
